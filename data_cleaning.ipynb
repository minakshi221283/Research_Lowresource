```python
import re
import pandas as pd

# Load CSV (replace 'input.csv' with your file path)
# Assuming the CSV has a column named 'text' that we want to clean
file = pd.read_excel("/content/Bhojpuri_combined (1).xlsx")
df = file['commentText']
df=pd.DataFrame(df)
df.head(10)
```

```python
import re
import pandas as pd


# Regex pattern for emojis/emoticons
EMOJI_PATTERN = re.compile(
    "["
    "\U0001F600-\U0001F64F"  # emoticons 😀-🙏
    "\U0001F300-\U0001F5FF"  # symbols & pictographs 🌍-💡
    "\U0001F680-\U0001F6FF"  # transport & map 🚀-🚦
    "\U0001F1E0-\U0001F1FF"  # flags 🇺🇸-🇮🇳
    "\U00002700-\U000027BF"  # dingbats ✨-➿
    "\U0001F900-\U0001F9FF"  # supplemental symbols 🤖-🦄
    "\U0001FA70-\U0001FAFF"  # extended pictographs 🛕-🧿
    "\u2764+"           # ❤ + variation selector ❤️
    "\u2665"
    "\u2764"
    "\U0001F5A4"
    "\u2764\uFE0F"# ♥ black heart suit
    "]+",
    flags=re.UNICODE
)

# Function to remove emojis from text
def remove_emojis(text):
    if pd.isna(text):  # handle missing values
        return "NULL "
    return EMOJI_PATTERN.sub(r'', str(text))

# Apply to the 'text' column (change if your column has a different name)
df["cleaned_text"] = df["commentText"].apply(remove_emojis)
df.head(25)







```

```python
df['cleaned_text'] = df['cleaned_text'].apply(lambda x: re.sub(r'[,\\n]+', '', str(x)))
df['cleaned_text'] = df['cleaned_text'].apply(lambda x: re.sub(r'[.,"\n]+', ' ', str(x)).strip())
df['cleaned_text'].replace('', np.nan, inplace=True)
df.dropna(inplace=True,axis=0)
df.head(25)
```

```python
# Save cleaned CSV
df['cleaned_text'].to_csv("cleaned_no_emojis.csv", index=False)
print("✅ Emoticons removed. Cleaned file saved as 'cleaned_no_emojis.csv'")
df.dropna(inplace=True)
```
