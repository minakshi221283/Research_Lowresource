```python
import pandas as pd
from googletrans import Translator
file_path = "/content/translated.csv"
output_path="/content/translated2.csv"

def translate_csv(file_path, output_path, target_language='hi'):
        df = pd.read_csv("/content/translated.csv")
        translator = Translator()
        translated_data = {}

        for column in df.columns:
            translated_data[column] = []
            for cell in df[column]:
                try:
                    if pd.notna(cell):
                        # Translate the cell's text
                        translated_text = translator.translate(str(cell), dest=target_language,src='en').text
                        translated_data[column].append(translated_text)
                    else:
                        # Keep empty cells as they are
                        translated_data[column].append(cell)
                except Exception as e:
                    print(f"Error translating cell: {cell}. Error: {e}")
                    translated_data[column].append(cell) # Append original cell in case of error

        translated_df = pd.DataFrame(translated_data)
        translated_df.to_csv(output_path, index=False)
        print(f"Translated file saved to {output_path}")

    # Example Usage:
translate_csv(file_path, output_path, 'hi')
```
