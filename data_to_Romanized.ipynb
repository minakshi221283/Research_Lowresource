```python
# ============================================================
# Bhojpuri -> Roman Script Batch Processor using uroman
# Supports:
#   - CSV input/output
#   - TXT input/output
#
# Install:
#   pip install pandas uroman
# ============================================================

!pip install uroman pandas

import pandas as pd
from uroman import Uroman
from pathlib import Path
import re

# ------------------------------------------------------------
# Initialize uroman
# ------------------------------------------------------------

u = Uroman()

# ------------------------------------------------------------
# Text normalization
# ------------------------------------------------------------

def normalize_text(text: str) -> str:
    text = str(text).strip().lower()
    text = re.sub(r"\s+", " ", text)
    return text

# ------------------------------------------------------------
# Romanization
# ------------------------------------------------------------

def romanize_bhojpuri(text: str) -> str:
    text = normalize_text(text)
    roman = u.romanize_string(text)
    roman = re.sub(r"\s+", " ", roman).strip()
    return roman

# ============================================================
# CSV PROCESSING
# ============================================================

def process_csv(
    input_csv,
    output_csv,
    input_column="text",
    output_column="romanized"
):
    """
    Reads a CSV file and adds a romanized column.
    """

    df = pd.read_excel("/content/Hatespeech_Bhojpuri_Hindi_english_final.xlsx")
    print("DataFrame columns:", df.columns)
    df = df[df['Language'] == "bho"]

    if input_column not in df.columns:
        raise ValueError(f"Column '{input_column}' not found.")

    df[output_column] = df[input_column].apply(romanize_bhojpuri)

    df.to_csv(output_csv, index=False)

    print(f"[DONE] Saved romanized CSV -> {output_csv}")
    print(df.head())


# ============================================================
# TXT PROCESSING
# ============================================================

def process_txt(input_txt, output_txt):
    """
    Reads line-by-line text file and writes romanized lines.
    """

    with open(input_txt, "r", encoding="utf-8") as f:
        lines = f.readlines()

    romanized_lines = [
        romanize_bhojpuri(line.strip())
        for line in lines
        if line.strip()
    ]

    with open(output_txt, "w", encoding="utf-8") as f:
        for line in romanized_lines:
            f.write(line + "\n")

    print(f"[DONE] Saved romanized TXT -> {output_txt}")


# ============================================================
# EXAMPLES
# ============================================================

if __name__ == "__main__":

    # Create a dummy input.txt for testing
    with open("input.txt", "w", encoding="utf-8") as f:
        f.write("का हाल बा\n")
        f.write("रउआ कहाँ जा तानी\n")

    # --------------------------------------------------------
    # Example CSV
    # --------------------------------------------------------
    #
    # input.csv
    #
    # text
    # का हाल बा
    # हम ठीक बानी
    #
    # --------------------------------------------------------

    process_csv(
        input_csv="input.csv",
        output_csv="output.csv",
        input_column="cleaned_text",
        output_column="romanized"
    )

    # --------------------------------------------------------
    # Example TXT
    # --------------------------------------------------------
    #
    # input.txt
    #
    # का हाल बा
    # रउआ कहाँ जा तानी
    #
    # --------------------------------------------------------

    process_txt(
        input_txt="input.txt",
        output_txt="output.txt"
    )
```
