```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df1=pd.read_excel('/content/Annotation_Bhojpuri.xlsx')
df2=pd.read_excel('/content/Annotation_Maithli.xlsx')
df3=pd.read_excel('/content/Annotation_awadhi.xlsx')
df4=pd.read_excel('/content/Annotation_braj.xlsx')
```

```python
plt.figure(figsize=(10,10))
plt.subplot(2,2,1)
plt.xlabel('Toxicity count')
plt.ylabel('Count')
plt.title('Bhojpuri')
sns.countplot(x=df1['Sentiment'],data=df1,hue=df1['Sentiment'],order=df1['Sentiment'].value_counts().index)
plt.subplot(2,2,2)
plt.xlabel('Toxicity count')
plt.ylabel('Count')
plt.title('Maithli')
sns.countplot(x=df2['Sentiment'],data=df2,hue=df2['Sentiment'],order=df2['Sentiment'].value_counts().index)
plt.subplot(2,2,3)
plt.xlabel('Toxicity count')
plt.ylabel('Count')
plt.title('Awadhi')
sns.countplot(x=df3['Sentiment'],data=df3,hue=df3['Sentiment'],order=df3['Sentiment'].value_counts().index)
plt.subplot(2,2,4)
plt.xlabel('Toxicity count')
plt.ylabel('Count')
plt.title('Braj')
sns.countplot(x=df4['Sentiment'],data=df4,hue=df4['Sentiment'],order=df4['Sentiment'].value_counts().index)
```
